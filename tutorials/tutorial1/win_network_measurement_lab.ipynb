<h2 style="color:#7dd3fc; margin-top:0;">🏓 Section 1 — <code style='background:#1e40af;padding:2px 8px;border-radius:4px;font-size:0.85em;'>ping</code>: Measuring Latency</h2>

<p style="color:#cbd5e1; margin-bottom:12px;"><code>ping</code> sends <strong>ICMP Echo Request</strong> packets to a host and waits for an <strong>ICMP Echo Reply</strong>. It operates at <strong>Layer 3 (Network Layer)</strong> — it does not use TCP or UDP.</p>

<p style="color:#cbd5e1; margin-bottom:8px;"><strong style="color:#7dd3fc;">Round Trip Time (RTT)</strong> = time from sending the request to receiving the reply. It is the sum of:</p>
<ul style="color:#94a3b8; margin:0; padding-left:20px;">
  <li><strong style="color:#e2e8f0;">Propagation delay</strong> — speed of light over physical distance</li>
  <li><strong style="color:#e2e8f0;">Transmission delay</strong> — time to push bits onto the wire</li>
  <li><strong style="color:#e2e8f0;">Queuing delay</strong> — wait time in router buffers under load</li>
  <li><strong style="color:#e2e8f0;">Processing delay</strong> — router/host CPU time</li>
</ul>
</div>

<h2 style="color:#7dd3fc; margin-top:0;">🗺️ Section 2 — <code style='background:#1e40af;padding:2px 8px;border-radius:4px;font-size:0.85em;'>tracert</code>: Discovering the Path</h2>

<p style="color:#cbd5e1; margin-bottom:12px;"><code>tracert</code> exploits the <strong>Time-To-Live (TTL)</strong> field in IP packets. Each router decrements TTL by 1. When TTL reaches 0, the router drops the packet and sends back an <strong>ICMP Time Exceeded</strong> message, revealing its IP address.</p>

<p style="color:#cbd5e1; margin-bottom:4px;">Traceroute sends probes with TTL = 1, 2, 3, … collecting replies to map each hop:</p>
<pre style="background:#0f2744; padding:10px; border-radius:6px; color:#e2e8f0; font-size:0.88em; margin:8px 0;">
TTL=1 → first router replies  (your gateway)
TTL=2 → second router replies (ISP core)
TTL=N → destination replies
</pre>
<p style="color:#94a3b8; font-size:0.88em; margin:0;">On Windows, <code>tracert</code> uses <strong>ICMP</strong> by default. Three probes are sent per TTL — the three RTT columns are the RTT for each probe.</p>
</div>

<h2 style="color:#7dd3fc; margin-top:0;">📊 Section 3 — <code style='background:#1e40af;padding:2px 8px;border-radius:4px;font-size:0.85em;'>iperf3</code>: Measuring Throughput</h2>

<p style="color:#cbd5e1; margin-bottom:12px;"><strong style="color:white;">Bandwidth</strong> is the <em>capacity</em> of a link (e.g., 100 Mbps fibre). <strong style="color:white;">Throughput</strong> is what you <em>actually achieve</em> — always ≤ bandwidth, limited by TCP slow-start, congestion, retransmissions, and bottleneck links.</p>

<p style="color:#cbd5e1; margin-bottom:8px;"><code>iperf3</code> uses a <strong>client-server</strong> model. We use the public OVH server to avoid needing your own:</p>
<pre style="background:#0f2744; padding:10px; border-radius:6px; color:#e2e8f0; font-size:0.88em; margin:8px 0;">
Server (already running): iperf3 -s          ← listens on port 5201
Client (you run):         iperf3 -c proof.ovh.net
</pre>
<p style="color:#fbbf24; font-size:0.87em; margin:0;">⚠️ If <code>proof.ovh.net</code> is busy, try <code>iperf.he.net</code> or <code>iperf3.speedtest.lu</code> as alternatives.</p>
</div>

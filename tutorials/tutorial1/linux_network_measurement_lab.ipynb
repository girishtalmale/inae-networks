<h2 style="color:#7dd3fc; margin-top:0;">🏓 Section 1 — <code style='background:#1e40af;padding:2px 8px;border-radius:4px;font-size:0.85em;'>ping</code>: Measuring Latency</h2>

<h3 style="color:#93c5fd; margin-bottom:6px;">What is <code>ping</code>?</h3>
<p style="color:#cbd5e1; margin-bottom:12px;"><code>ping</code> is one of the oldest and most universally useful network tools — it has been around since 1983 and is installed on virtually every operating system. It sends <strong>ICMP Echo Request</strong> packets to a host and waits for an <strong>ICMP Echo Reply</strong>. It operates at <strong>Layer 3 (Network Layer)</strong> — it does not use TCP or UDP, so it bypasses applications entirely and speaks directly to the IP stack of the destination.</p>

<h3 style="color:#93c5fd; margin-bottom:6px;">What does it actually measure?</h3>
<p style="color:#cbd5e1; margin-bottom:8px;"><strong style="color:#7dd3fc;">Round Trip Time (RTT)</strong> = time from sending the request to receiving the reply. It is the sum of:</p>
<ul style="color:#94a3b8; margin:0 0 12px 0; padding-left:20px;">
  <li><strong style="color:#e2e8f0;">Propagation delay</strong> — speed of light over physical distance</li>
  <li><strong style="color:#e2e8f0;">Transmission delay</strong> — time to push bits onto the wire</li>
  <li><strong style="color:#e2e8f0;">Queuing delay</strong> — wait time in router buffers under load</li>
  <li><strong style="color:#e2e8f0;">Processing delay</strong> — router/host CPU time</li>
</ul>

<h3 style="color:#93c5fd; margin-bottom:6px;">Where is <code>ping</code> used in real life?</h3>
<ul style="color:#94a3b8; margin:0 0 14px 0; padding-left:20px;">
  <li><strong style="color:#e2e8f0;">Reachability check</strong> — "Is that server even online?" — the first thing any sysadmin runs.</li>
  <li><strong style="color:#e2e8f0;">Latency baseline</strong> — Game servers, VoIP systems, and trading platforms all care deeply about RTT. A gamer will ping a game server to decide which region to connect to.</li>
  <li><strong style="color:#e2e8f0;">Jitter diagnosis</strong> — The <code>mdev</code> field (mean deviation) measures how <em>inconsistent</em> latency is. High mdev → choppy video calls, stuttering games.</li>
  <li><strong style="color:#e2e8f0;">Network monitoring</strong> — Production monitoring systems (Nagios, Zabbix, Datadog) continuously ping services and alert on loss or latency spikes.</li>
  <li><strong style="color:#e2e8f0;">ISP troubleshooting</strong> — Support engineers will ask you to paste a ping output to diagnose if packet loss is local or upstream.</li>
</ul>

<p style="color:#fbbf24; font-size:0.9em; margin:0;">💡 <strong>Why this is interesting:</strong> A single 64-byte packet bounces across potentially thousands of kilometres of fibre, through dozens of routers, and comes back — in under 5 milliseconds for a nearby CDN node. The fact that this works reliably at all, billions of times per second across the Internet, is a remarkable engineering achievement.</p>
</div>

  <h2 style="color:#7dd3fc; margin-top:0;">🗺️ Section 2 — <code style='background:#1e40af;padding:2px 8px;border-radius:4px;font-size:0.85em;'>traceroute</code>: Discovering the Path</h2>

  <h3 style="color:#93c5fd; margin-bottom:6px;">What is <code>traceroute</code>?</h3>
  <p style="color:#cbd5e1; margin-bottom:12px;"><code>traceroute</code> is a network diagnostic tool that shows the <strong>sequence of routers (hops)</strong> a packet passes through to reach a destination. It works by sending probes with increasing <strong>TTL (Time-To-Live)</strong> values. Each router decreases the TTL by 1; when TTL becomes 0, the packet is dropped and the router sends back an <strong>ICMP Time Exceeded</strong> message, revealing its presence and IP address.</p>

  <h3 style="color:#93c5fd; margin-bottom:6px;">How does it work?</h3>
  <p style="color:#cbd5e1; margin-bottom:8px;"><strong style="color:#7dd3fc;">TTL-based hop discovery</strong> lets <code>traceroute</code> map the route hop by hop:</p>
  <pre style="background:#0f2744; padding:10px; border-radius:6px; color:#e2e8f0; font-size:0.88em; margin:8px 0;">
TTL=1 → first router replies
TTL=2 → second router replies
TTL=3 → third router replies
...
TTL=N → destination reached
  </pre>
  <p style="color:#94a3b8; font-size:0.88em; margin:0;">On Linux, the default probes are usually <strong>UDP</strong>. Use <code>-I</code> to send <strong>ICMP</strong> probes instead. Each hop often shows three RTT values because three probes are sent per TTL.</p>

  <h3 style="color:#93c5fd; margin-bottom:6px;">What does it actually measure?</h3>
  <ul style="color:#94a3b8; margin:0 0 12px 0; padding-left:20px;">
    <li><strong style="color:#e2e8f0;">Path visibility</strong> — which routers / networks are traversed</li>
    <li><strong style="color:#e2e8f0;">Hop-by-hop delay</strong> — RTT to each hop along the route</li>
    <li><strong style="color:#e2e8f0;">Routing behavior</strong> — changes in path, load balancing, or asymmetry</li>
    <li><strong style="color:#e2e8f0;">Network reachability</strong> — where packets stop or become blocked</li>
  </ul>

  <h3 style="color:#93c5fd; margin-bottom:6px;">Where is <code>traceroute</code> used in real life?</h3>
  <ul style="color:#94a3b8; margin:0 0 14px 0; padding-left:20px;">
    <li><strong style="color:#e2e8f0;">Troubleshooting connectivity</strong> — helps find where packets are delayed or dropped.</li>
    <li><strong style="color:#e2e8f0;">ISP diagnostics</strong> — support teams use it to check whether a problem is in the local network, ISP core, or destination side.</li>
    <li><strong style="color:#e2e8f0;">Routing analysis</strong> — network engineers inspect route changes, peering paths, and unexpected detours.</li>
    <li><strong style="color:#e2e8f0;">Latency debugging</strong> — when RTT is high, traceroute helps identify which hop introduces the jump.</li>
    <li><strong style="color:#e2e8f0;">Security and filtering checks</strong> — `* * *` responses may indicate firewall rules, ICMP blocking, or rate limiting.</li>
  </ul>

  <p style="color:#fbbf24; font-size:0.9em; margin:0;">💡 <strong>Why this is interesting:</strong> With just a few tiny probe packets, you can reveal the hidden route your traffic takes across the Internet — from your local gateway, through ISP routers, and often across multiple autonomous systems before reaching the destination.</p>
</div>

### Task 2.2 — Understand `* * *` and Multiple IPs

<div style="background: linear-gradient(135deg, #0f172a 0%, #1e3a5f 100%); border-radius: 10px; padding: 16px; font-family: monospace; font-size: 0.87em; color: #e2e8f0; margin: 8px 0;">
<strong style="color:#7dd3fc;">📖 What does <code>* * *</code> mean?</strong>
<p style="font-family:'Segoe UI',sans-serif; color:#cbd5e1; margin:10px 0 6px 0;">When traceroute shows <code>* * *</code> for a hop, it means <strong>no ICMP reply arrived within the timeout window</strong> for all three probes sent at that TTL. This is commonly misread as "the packet was dropped here" — but that is often wrong:</p>
<ul style="font-family:'Segoe UI',sans-serif; color:#94a3b8; margin:0; padding-left:20px;">
  <li><strong style="color:#e2e8f0;">Router rate-limits ICMP</strong> — Many routers intentionally deprioritise sending ICMP Time Exceeded messages (generating them costs CPU). The packet passes through fine, but no reply is sent. Notice: if subsequent hops <em>do</em> respond, the destination is still reachable.</li>
  <li><strong style="color:#e2e8f0;">Firewall silently drops</strong> — A network operator may block ICMP responses on a router interface as a security policy, while forwarding the data traffic normally.</li>
  <li><strong style="color:#e2e8f0;">Asymmetric routing</strong> — The return ICMP packet takes a different path that is blocked, even if the forward path is fine.</li>
  <li><strong style="color:#e2e8f0;">Genuine loss</strong> — Only when <em>all</em> subsequent hops also show <code>* * *</code> and the destination itself is unreachable does it indicate actual packet loss.</li>
</ul>
<p style="font-family:'Segoe UI',sans-serif; color:#fbbf24; margin:10px 0 0 0; font-size:0.88em;">⚠️ <strong>Rule of thumb:</strong> <code>* * *</code> followed by a responding hop = the router is just silent, not broken. Worry only when <code>* * *</code> continues to the end.</p>
</div>

  <h2 style="color:#7dd3fc; margin-top:0;">📊 Section 3 — <code style='background:#1e40af;padding:2px 8px;border-radius:4px;font-size:0.85em;'>iperf3</code>: Measuring Throughput</h2>

  <h3 style="color:#93c5fd; margin-bottom:6px;">What is <code>iperf3</code>?</h3>
  <p style="color:#cbd5e1; margin-bottom:12px;"><code>iperf3</code> is a network performance measurement tool used to measure <strong>throughput</strong>, <strong>packet loss</strong>, and <strong>jitter</strong>. Unlike <code>ping</code>, which measures latency using tiny ICMP packets, <code>iperf3</code> actively generates large volumes of TCP or UDP traffic between a <strong>client</strong> and a <strong>server</strong> to test how much data the network can actually carry.</p>

  <h3 style="color:#93c5fd; margin-bottom:6px;">Bandwidth vs Throughput</h3>
  <p style="color:#cbd5e1; margin-bottom:8px;"><strong style="color:white;">Bandwidth</strong> is the theoretical <em>capacity</em> of a link (e.g., 100 Mbps fibre). <strong style="color:white;">Throughput</strong> is the <em>actual achieved transfer rate</em>. Throughput is almost always lower due to:</p>
  <ul style="color:#94a3b8; margin:0 0 12px 0; padding-left:20px;">
    <li><strong style="color:#e2e8f0;">TCP slow-start</strong> — connections ramp up gradually</li>
    <li><strong style="color:#e2e8f0;">Congestion control</strong> — TCP reduces sending rate under loss</li>
    <li><strong style="color:#e2e8f0;">Retransmissions</strong> — lost packets must be resent</li>
    <li><strong style="color:#e2e8f0;">Bottleneck links</strong> — the slowest link limits end-to-end throughput</li>
  </ul>

  <h3 style="color:#93c5fd; margin-bottom:6px;">How does <code>iperf3</code> work?</h3>
  <p style="color:#cbd5e1; margin-bottom:8px;"><code>iperf3</code> uses a <strong>client-server model</strong>. One machine runs the server and listens on port <code>5201</code>; the other machine runs the client and generates traffic.</p>

  <pre style="background:#0f2744; padding:10px; border-radius:6px; color:#e2e8f0; font-size:0.88em; margin:8px 0;">
Server (already running): iperf3 -s
Client (you run):         iperf3 -c proof.ovh.net
  </pre>

  <p style="color:#fbbf24; font-size:0.87em; margin:0 0 14px 0;">⚠️ If <code>proof.ovh.net</code> is busy, try <code>iperf.he.net</code> or <code>iperf3.speedtest.lu</code>.</p>

  <h3 style="color:#93c5fd; margin-bottom:6px;">What does it actually measure?</h3>
  <ul style="color:#94a3b8; margin:0 0 12px 0; padding-left:20px;">
    <li><strong style="color:#e2e8f0;">TCP throughput</strong> — achievable end-to-end transfer speed</li>
    <li><strong style="color:#e2e8f0;">UDP packet loss</strong> — percentage of dropped packets</li>
    <li><strong style="color:#e2e8f0;">Jitter</strong> — variation in packet arrival timing</li>
    <li><strong style="color:#e2e8f0;">Retransmissions</strong> — packets resent due to loss or congestion</li>
    <li><strong style="color:#e2e8f0;">Parallel stream behavior</strong> — performance under multiple simultaneous flows</li>
  </ul>

  <h3 style="color:#93c5fd; margin-bottom:6px;">Where is <code>iperf3</code> used in real life?</h3>
  <ul style="color:#94a3b8; margin:0 0 14px 0; padding-left:20px;">
    <li><strong style="color:#e2e8f0;">ISP speed diagnostics</strong> — engineers test whether a customer actually receives expected throughput.</li>
    <li><strong style="color:#e2e8f0;">Data center benchmarking</strong> — operators validate server-to-server link performance.</li>
    <li><strong style="color:#e2e8f0;">WiFi performance testing</strong> — compare throughput across rooms, floors, or access points.</li>
    <li><strong style="color:#e2e8f0;">Congestion analysis</strong> — identify throughput collapse caused by packet loss or RTT.</li>
    <li><strong style="color:#e2e8f0;">Research and experimentation</strong> — widely used in networking papers and systems experiments for evaluating transport protocols.</li>
  </ul>

  <p style="color:#fbbf24; font-size:0.9em; margin:0;">💡 <strong>Why this is interesting:</strong> Modern networks may advertise gigabit speeds, but the throughput you actually achieve depends on protocol behavior, congestion, RTT, packet loss, and routing conditions. <code>iperf3</code> exposes the real performance of the network rather than just the advertised link capacity.</p>

<p style="color:#7dd3fc; font-size:0.85em; font-weight:700; letter-spacing:1px; text-transform:uppercase; margin:0 0 6px 0;">▶ Why didn't a single TCP stream saturate the link?</p>
<p style="color:#cbd5e1; margin:0; font-size:0.93em;">
TCP uses a <strong>congestion window (cwnd)</strong> to control how much data it injects into the network at once. It starts small (slow start) and grows — but it is a <em>per-flow</em> limit. A single stream negotiates with the network for its fair share of bandwidth, backing off every time it detects congestion (packet loss or ECN marks).<br><br>
The result: on a high-latency path, one TCP stream may never fully open its window before the round-trip time "forces" it to wait for acknowledgements — leaving bandwidth on the table. <strong>What happens if we open multiple independent streams in parallel?</strong> Each one competes independently, and together they can fill the pipe that one stream couldn't. That's exactly what Task 3.2 tests.
</p>

<p style="color:#f59e0b; font-size:0.85em; font-weight:700; letter-spacing:1px; text-transform:uppercase; margin:0 0 6px 0;">▶ Up Next — Section 3.3: A fundamentally different transport</p>
<p style="color:#cbd5e1; margin:0; font-size:0.93em;">
Everything in Tasks 3.1 and 3.2 used <strong>TCP</strong> — a transport protocol that <em>fights against congestion</em>. When TCP detects a dropped packet, it slows down. When it sees the network is clear, it ramps up. The <code>cwnd</code>, retransmissions, and the ramp-up curve you observed are all TCP trying to be a "good citizen" on a shared network.<br><br>
<strong>UDP does none of that.</strong> There is no congestion window, no retransmissions, no flow control. UDP sends packets at whatever rate you tell it to — and if the network can't keep up, packets are simply <em>dropped silently</em>. This sounds reckless, but it's actually the right choice for many real-world applications. Task 3.3 lets you feel the difference directly — and compare the two transport protocols side by side.
</p>
